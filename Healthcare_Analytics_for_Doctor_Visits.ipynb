```python
import matplotlib.pyplot as plt
import seaborn as sns

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# 1. Distribution of Doctor Visits (Discrete Count)
sns.countplot(data=df, x='visits', hue='visits', legend=False, ax=axes[0], palette='viridis')
axes[0].set_title('Distribution of Doctor Visits')
axes[0].set_xlabel('Number of Visits')
axes[0].set_ylabel('Count')

# 2. Distribution of Age (Continuous Variable)
sns.histplot(data=df, x='age', kde=True, ax=axes[1], color='skyblue')
axes[1].set_title('Distribution of Patient Age')
axes[1].set_xlabel('Age (Scaled/Normalized)')
axes[1].set_ylabel('Count')

# 3. Proportion of Gender (Categorical Variable)
sns.countplot(data=df, x='gender', hue='gender', legend=False, ax=axes[2], palette='pastel')
axes[2].set_title('Gender Distribution')
axes[2].set_xlabel('Gender')
axes[2].set_ylabel('Count')

plt.tight_layout()
plt.show()
```

```python
gender_analysis = (
    df.groupby("gender")["visits"]
    .agg(["count", "mean", "median"])
    .reset_index()
)

display(gender_analysis)

sns.barplot(
    data=gender_analysis,
    x="gender",
    y="mean",
    color="teal"
)

plt.title("Average Doctor Visits by Gender")
plt.xlabel("Gender")
plt.ylabel("Mean recorded visits")
```

```python
illness_analysis = (
    df.groupby("illness")["visits"]
    .agg(["count", "mean"])
    .reset_index()
)

display(illness_analysis)

sns.lineplot(
    data=illness_analysis,
    x="illness",
    y="mean",
    marker="o"
)

plt.title("Illness Score vs. Average Doctor Visits")
plt.xlabel("Illness score")
plt.ylabel("Mean recorded visits")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
# Scatter plot displaying Income vs Age, color-coded by Gender, with size representing number of Visits
sns.scatterplot(
    data=df,
    x='age',
    y='income',
    hue='gender',
    size='visits',
    sizes=(20, 200),
    alpha=0.6,
    palette='Set1'
)

plt.title('Multivariate Analysis: Age vs Income by Gender and Doctor Visits')
plt.xlabel('Age (Scaled/Normalized)')
plt.ylabel('Income')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()
```
